# Encapsulation in Python

### Simple idea

> **Encapsulation = Keep data and the methods that work on that data together, and control how that data is accessed or changed.**

We will use a **Bank Account** example because it makes the idea easy to understand.

### Why do we need it?

Imagine a bank account:

```text
Bank Account
├── account holder
├── balance
├── deposit()
└── withdraw()
```

The `balance` is important data.

We usually do **not** want anyone to freely change it like this:

```python
account.balance = -1000000
```

Instead, we want changes to happen through controlled methods such as:

```python
account.deposit(500)
account.withdraw(200)
```

That is where **encapsulation** helps.


## 1. Without Encapsulation

Here the balance is public, so anyone can change it directly.


````python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance


account = BankAccount("Alice", 1000)

print("Starting balance:", account.balance)

# Anyone can directly change the balance
account.balance = -5000

print("Changed balance:", account.balance)
````

### What is the problem?

The program allowed this:

```python
account.balance = -5000
```

There was **no validation** and no control over how the balance changed.

For important data such as:

- bank balance
- password
- salary
- marks
- account status

we may want to **control access and updates**.


## 2. Encapsulation

We can hide or restrict direct access to internal data and provide methods to work with it safely.

```text
Outside Code
     │
     │ deposit(500)
     │ withdraw(200)
     ▼
┌─────────────────┐
│   BankAccount   │
│                 │
│  __balance      │  ← internal data
│                 │
│  deposit()      │
│  withdraw()     │
│  get_balance()  │
└─────────────────┘
```

> Instead of changing the balance directly, we go through controlled methods.


````python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance   # private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawal successful")

    def get_balance(self):
        return self.__balance


account = BankAccount("Alice", 1000)

account.deposit(500)
account.withdraw(200)

print("Balance:", account.get_balance())
````

## 3. How Do We Achieve Encapsulation in Python?

Python commonly uses three levels:

| Type | Example | Meaning |
|---|---|---|
| **Public** | `self.name` | Can be accessed normally |
| **Protected** | `self._age` | Intended for internal/subclass use |
| **Private** | `self.__balance` | Direct access is discouraged using name mangling |

### Easy memory

```text
name       → Public
_name      → Protected
__name     → Private
```

> Python is different from languages like Java or C++.  
> Protected/private access is based partly on **convention**, and double underscore uses **name mangling**.


## 4. Public Members

Public members can be accessed directly from outside the class.


````python
class Student:

    def __init__(self, name):
        self.name = name   # public


student = Student("Alice")

print(student.name)

student.name = "Maya"

print(student.name)
````

### When is public access fine?

Use public attributes when direct access is simple and safe.

Examples:

- student's name
- product title
- city
- course name

Not every variable needs to be hidden.


## 5. Protected Members

A protected member uses **one underscore**:

```python
self._age
```

It means:

> “This is intended mainly for use inside this class or its subclasses.”

Python does **not strictly block** outside access.


````python
class Person:

    def __init__(self, name, age):
        self.name = name
        self._age = age   # protected


class Student(Person):

    def show_age(self):
        print("Age:", self._age)


student = Student("Alice", 20)

student.show_age()

# This still works in Python,
# but _age is intended to be treated as internal.
print(student._age)
````

## 6. Private Members

A private member uses **two underscores**:

```python
self.__balance
```

This makes direct access harder and tells other programmers:

> “Do not access this directly. Use the class methods.”

Python internally changes the name using **name mangling**.


````python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance


account = BankAccount(1000)

# This will raise an AttributeError if uncommented:
# print(account.__balance)

print("Direct access to __balance is not available in the normal way.")
````

## 7. Getter Methods

A **getter** gives controlled read access to private data.


````python
class Employee:

    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary


employee = Employee(50000)

print(employee.get_salary())
````

## 8. Setter Methods

A **setter** gives controlled update access.

This is useful because we can validate the new value before changing the data.


````python
class Employee:

    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
            print("Salary updated")
        else:
            print("Salary must be greater than 0")


employee = Employee(50000)

employee.set_salary(60000)
print(employee.get_salary())

employee.set_salary(-100)
print(employee.get_salary())
````

## 9. Why Are Getters and Setters Useful?

Without validation:

```python
employee.salary = -50000
```

With a setter:

```python
employee.set_salary(-50000)
```

the class can reject invalid values.

### This gives us:

- safer updates
- validation
- better control
- easier maintenance
- protection from accidental changes


## 10. Real Use Case: Bank Account

This example brings everything together.

We want:

- `owner` → public
- `__balance` → private
- `deposit()` → safely add money
- `withdraw()` → safely remove money
- `get_balance()` → safely view balance


````python
class BankAccount:

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"${amount} deposited")
        else:
            print("Deposit must be greater than 0")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal must be greater than 0")
        elif amount > self.__balance:
            print("Not enough money")
        else:
            self.__balance -= amount
            print(f"${amount} withdrawn")

    def get_balance(self):
        return self.__balance


account = BankAccount("Alice", 1000)

print("Owner:", account.owner)
print("Starting balance:", account.get_balance())

account.deposit(500)
account.withdraw(300)

print("Final balance:", account.get_balance())
````

## 11. Encapsulation Is More Than “Making Variables Private”

The real idea is:

> **The class controls how its internal data is used and changed.**

For example, `withdraw()` protects the balance by checking:

```text
Is the amount valid?
        ↓
Is there enough money?
        ↓
Only then update balance
```

So encapsulation helps protect the **rules of the object**, not just hide variables.


## 12. Where Can Encapsulation Be Helpful?

### Bank Account
Protect `balance` and allow changes only through `deposit()` / `withdraw()`.

### Employee
Protect `salary` and validate salary changes.

### Student
Protect `marks` and only allow values between `0` and `100`.

### User Account
Protect password-related data and control how it is updated.

### Product
Protect price and reject negative prices.

> Use encapsulation when data should not be changed freely without rules.
