# Polymorphism in Python

### Simple idea

> **Polymorphism = Same action, different behavior.**

The word means:

```text
Poly  → many
Morph → forms
```

So:

> **Polymorphism = many forms**

For example, imagine different notification types:

```text
Email → send()
SMS   → send()
Push  → send()
```

The method name is the same:

```python
send()
```

But each object performs the action differently.

That is the main idea of **polymorphism**.


## 1. Why Do We Need Polymorphism?

Imagine an application that sends notifications.

We may have:

- Email Notification
- SMS Notification
- Push Notification

Without a common method name, we might create:

```text
send_email()
send_sms()
send_push_notification()
```

Then the rest of the program must remember which function belongs to which notification type.

Wouldn't it be easier if every notification simply had:

```python
send()
```

Then:

```text
Email.send() → sends an email
SMS.send()   → sends an SMS
Push.send()  → sends a push notification
```

> Same method name.  
> Different objects.  
> Different behavior.


## 2. Without Polymorphism

Here every class uses a different method name.


````python
class Email:

    def send_email(self):
        print("Sending an email")


class SMS:

    def send_sms(self):
        print("Sending an SMS")


email = Email()
sms = SMS()

email.send_email()
sms.send_sms()
````

### What is the problem?

Both objects perform the same general action:

> **Send a notification**

But the program must remember:

```text
Email → send_email()
SMS   → send_sms()
```

If we add more types:

```text
Push Notification
WhatsApp
Slack
Teams
```

we may end up remembering many different method names.

Polymorphism lets us use one common action:

```python
send()
```


## 3. With Polymorphism

Give every class the same method name:

```python
send()
```

Each class provides its own behavior.


````python
class Email:

    def send(self):
        print("Sending an email")


class SMS:

    def send(self):
        print("Sending an SMS")


class PushNotification:

    def send(self):
        print("Sending a push notification")


email = Email()
sms = SMS()
push = PushNotification()

email.send()
sms.send()
push.send()
````

### What happened?

We called:

```python
send()
```

three times.

But the result changed depending on the object:

```text
Email object → Email version of send()
SMS object   → SMS version of send()
Push object  → Push version of send()
```

> **Same method name → different behavior**

That is polymorphism.


## 4. One Function Can Work With Many Objects

This is where polymorphism becomes very useful.

Because all notification objects have a `send()` method, one function can work with all of them.


````python
class Email:

    def send(self):
        print("Email sent")


class SMS:

    def send(self):
        print("SMS sent")


class PushNotification:

    def send(self):
        print("Push notification sent")


def send_notification(notification):
    notification.send()


email = Email()
sms = SMS()
push = PushNotification()

send_notification(email)
send_notification(sms)
send_notification(push)
````

### Why is this powerful?

The function:

```python
send_notification()
```

does not need separate code for every notification type.

It simply says:

```python
notification.send()
```

As long as the object has a `send()` method, the function can use it.

```text
send_notification()
        │
        └── send()
             │
             ├── Email behavior
             ├── SMS behavior
             └── Push behavior
```

This makes code:

- simpler
- reusable
- easier to extend




```
             ABSTRACTION
                 ↓
        "Every notification
          should have send()"
                 ↓
              send()
                 ↓
        ┌────────┼────────┐
        ↓        ↓        ↓
      Email     SMS      Push
        ↓        ↓        ↓
     email      text     push
     sent       sent     sent
        └────────┬────────┘
                 ↑
             POLYMORPHISM
       Same send(), different behavior
```

Easy way to remember
- Abstraction = What should be done?
- Polymorphism = How does each object do it differently?

## 6. Polymorphism Through Inheritance

A very common form of polymorphism happens with **inheritance**.

A parent class defines a method.

Child classes provide their own version of that method.

This is called **method overriding**.


````python
class Animal:

    def sound(self):
        return "Some animal sound"


class Dog(Animal):

    def sound(self):
        return "Woof"


class Cat(Animal):

    def sound(self):
        return "Meow"


animal = Animal()
dog = Dog()
cat = Cat()

print(animal.sound())
print(dog.sound())
print(cat.sound())
````

## 7. Method Overriding

Here the parent class has:

```python
sound()
```

The child classes also define:

```python
sound()
```

Their versions **override** the parent's version.

```text
Animal.sound()
      │
      ├── Dog.sound() → "Woof"
      └── Cat.sound() → "Meow"
```

When the program runs, Python chooses the correct method based on the actual object.

This is commonly called:

> **Runtime polymorphism**


````python
animals = [
    Dog(),
    Cat(),
    Animal()
]

for animal in animals:
    print(animal.sound())
````

## 9. Python's Built-in Functions Are Polymorphic Too

You have already used polymorphism without realizing it.

For example:

```python
len()
```

works with different types of objects.


````python
print(len("Python"))

print(len([10, 20, 30, 40]))

print(len({"name": "Alice", "age": 20}))
````

We used the same function:

```python
len()
```

with:

- a string
- a list
- a dictionary

The function works with different object types.

Another example is:

```python
max()
```


````python
print(max(10, 50, 20))

print(max("apple", "banana", "mango"))
````

> Same function name.  
> Different kinds of data.  
> Appropriate behavior for that data.

That is also an example of polymorphism.


## 10. Operators Can Be Polymorphic Too

The same operator can behave differently depending on the data.

Consider:

```python
+
```


````python
# Numbers → addition
print(10 + 20)

# Strings → joining / concatenation
print("Hello " + "World")

# Lists → joining lists
print([1, 2] + [3, 4])
````

The operator is always:

```text
+
```

But its behavior changes:

```text
Numbers → addition
Strings → concatenation
Lists   → list concatenation
```

This is called **operator polymorphism**.

You may also hear this described as **operator overloading**.


## 11. What About Method Overloading?

Some languages allow multiple methods with exactly the same name but different parameter lists.

- Python does not natively support traditional method overloading
- In languages like Java or C++, you can write multiple methods with the exact same name but different signatures (different number or types of parameters).
- In Python, if you define multiple functions or methods with the same name, the latest definition simply overwrites the previous ones.

For example:

```text
multiply(a, b)
multiply(a, b, c)
multiply(a, b, c, d)
```

Python does **not** support traditional method overloading in the same way as languages such as Java or C++.

Instead, we can often achieve similar behavior using:

- default arguments
- `*args`
- `**kwargs`

For this course, the important idea is:

> **Python can make one method flexible enough to accept different inputs.**


````python
class Calculator:

    def multiply(self, *numbers):
        result = 1

        for number in numbers:
            result *= number

        return result


calculator = Calculator()

print(calculator.multiply(2, 3))
print(calculator.multiply(2, 3, 4))
print(calculator.multiply(2, 3, 4, 5))
````

Here we still use one method:

```python
multiply()
```

but it can handle different numbers of arguments.

This is **overloading-like behavior**, not traditional compile-time method overloading.


## 14. Polymorphism vs Abstraction

These concepts often work together, but they are different.

| Polymorphism | Abstraction |
|---|---|
| Same action, different behavior | Show what is needed, hide how it works |
| Focuses on many implementations | Focuses on a simple interface |
| Example: `Dog.sound()` / `Cat.sound()` | Example: every payment has `pay()` |
| Python chooses behavior based on object | User does not need internal details |

### Easy memory

> **Polymorphism = Many behaviors**

> **Abstraction = Hide complexity**


## 15. Polymorphism vs Inheritance

Inheritance and polymorphism are related, but they are not the same thing.

### Inheritance

```text
Dog IS-A Animal
Cat IS-A Animal
```

A child class receives features from a parent class.

### Polymorphism

```text
Dog.sound() → Woof
Cat.sound() → Meow
```

The same method behaves differently for different objects.

> **Inheritance = relationship between classes**

> **Polymorphism = same interface, different behavior**

Polymorphism can use inheritance, but Python can also achieve polymorphism through **duck typing** without inheritance.
