# Inheritance and Composition in Python

Both inheritance and composition help us build relationships between classes and reuse code.

### Easy memory

```text
Inheritance  → IS-A
Composition  → HAS-A
```

Examples:

```text
Dog IS-A Animal        → Inheritance
Car HAS-A Engine       → Composition
```

We will learn both together because the difference becomes much easier to see.


# Part 1 — Inheritance

## 1. Why Do We Need Inheritance?

Imagine we are building a system with:

- Cars
- Bikes
- Trucks

All of them may share common features:

```text
Vehicle
├── brand
├── start()
└── stop()
```

Without inheritance, we may repeat the same code in every class.


````python
# WITHOUT inheritance
# Notice the repeated code.

class Car:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} car started")


class Bike:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} bike started")


car = Car("Toyota")
bike = Bike("Honda")

car.start()
bike.start()
````

### What is the problem?

`Car` and `Bike` both need:

- `brand`
- `start()`

We are repeating similar code.

If many classes share the same features, we can move those common features into one **parent class**.

That is where inheritance helps.


## 2. What Is Inheritance?

> **Inheritance allows one class to reuse and extend the attributes and methods of another class.**

The existing class is called the:

- **Parent class**
- **Base class**
- **Superclass**

The class that inherits from it is called the:

- **Child class**
- **Derived class**
- **Subclass**

### Simple picture

```text
          Vehicle
        Parent Class
             │
       ┌─────┴─────┐
       ↓           ↓
      Car         Bike
    Child         Child
```

> `Car` **IS-A** `Vehicle`  
> `Bike` **IS-A** `Vehicle`


## 3. How Do We Create Inheritance?

Put the parent class name inside parentheses:

```python
class Child(Parent):
    ...
```

Example:

```python
class Car(Vehicle):
    ...
```


````python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} vehicle started")


class Car(Vehicle):
    pass


car = Car("Toyota")

# Car inherited both brand and start()
print(car.brand)
car.start()
````

### What happened?

`Car` did not define:

```python
__init__()
start()
```

But it can still use them because it inherited them from `Vehicle`.

```text
Vehicle
├── brand
└── start()
     │
     ↓ inherited by
    Car
```

> **Inheritance lets us write common code once and reuse it in child classes.**


## 4. A Child Class Can Add Its Own Features

A child does not only inherit.

It can also add new attributes and methods that belong only to that child.


````python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"{self.brand} vehicle started")


class Car(Vehicle):

    def open_trunk(self):
        print("Car trunk opened")


car = Car("Toyota")

# Inherited method
car.start()

# Child's own method
car.open_trunk()
````

Think of it like this:

```text
Vehicle
├── brand
└── start()
     │
     ▼
Car
├── inherits brand
├── inherits start()
└── adds open_trunk()
```


## 5. Using `super()`

Sometimes the child needs its own `__init__()` but still wants to reuse the parent's constructor.

We use:

```python
super().__init__(...)
```

### Easy memory

> **`super()` = use something from the parent class**


````python
class Vehicle:

    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):

    def __init__(self, brand, model):
        # Reuse the parent's constructor
        super().__init__(brand)

        # Add child-specific data
        self.model = model


car = Car("Toyota", "Corolla")

print(car.brand)
print(car.model)
````

### Why use `super()`?

Without `super()`, we may rewrite:

```python
self.brand = brand
```

inside every child.

With `super()`:

```python
super().__init__(brand)
```

we reuse the parent's setup.

This helps reduce repeated code.


## 6. Method Overriding

A child class can replace an inherited method with its own version.

This is called **method overriding**.


````python
class Vehicle:

    def move(self):
        print("Vehicle is moving")


class Car(Vehicle):

    def move(self):
        print("Car is driving")


class Boat(Vehicle):

    def move(self):
        print("Boat is sailing")


car = Car()
boat = Boat()

car.move()
boat.move()
````

### What happened?

The parent already had:

```python
move()
```

But the child classes provided their own versions.

```text
Vehicle.move()
      │
      ├── Car.move()  → driving
      └── Boat.move() → sailing
```

This is also closely connected to **polymorphism**:

> same method name, different behavior.


## 7. Calling the Parent Version From an Overridden Method

Sometimes we want the child method to do something extra **after** the parent's behavior.

We can call the parent version using `super()`.


````python
class Vehicle:

    def start(self):
        print("Basic vehicle checks completed")


class ElectricCar(Vehicle):

    def start(self):
        super().start()
        print("Battery system activated")


car = ElectricCar()
car.start()
````

## 8. Benefits of Inheritance

Inheritance can help with:

- **Code reuse** → write common behavior once
- **Less repetition** → avoid copying the same methods
- **Organization** → related classes form a hierarchy
- **Extension** → children can add new features
- **Customization** → children can override behavior
- **Maintenance** → common changes can often be made in the parent

Example:

```text
Employee
├── Manager
├── Developer
└── Designer
```

All employees may share:

```text
name
employee_id
login()
```

while each role adds its own behavior.


## 9. Common Types of Inheritance

You do not need to memorize all of these today, but you should recognize the patterns.

### 1. Single Inheritance

One parent → one child

```text
Animal
  │
  ▼
 Dog
```

---

### 2. Multilevel Inheritance

A child becomes the parent of another class.

```text
Animal
  │
  ▼
Mammal
  │
  ▼
 Dog
```

---

### 3. Hierarchical Inheritance

One parent → multiple children

```text
       Vehicle
      /       \
    Car       Bike
```

---

### 4. Multiple Inheritance

One child inherits from multiple parents.

```text
Camera     Phone
    \       /
     \     /
    Smartphone
```

---

### 5. Hybrid Inheritance

A combination of inheritance patterns.

> For beginners, focus mainly on **single** and **hierarchical** inheritance.


## 10. Single Inheritance Example


````python
class Animal:

    def eat(self):
        print("Animal is eating")


class Dog(Animal):

    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()   # inherited
dog.bark()  # own method
````

## 11. Multilevel Inheritance Example


````python
class Animal:

    def breathe(self):
        print("Animal breathes")


class Mammal(Animal):

    def walk(self):
        print("Mammal walks")


class Dog(Mammal):

    def bark(self):
        print("Dog barks")


dog = Dog()

dog.breathe()
dog.walk()
dog.bark()
````

## 12. Multiple Inheritance — Simple Introduction

Python also allows a child to inherit from more than one parent.

```python
class Child(Parent1, Parent2):
    ...
```

Keep this simple for now.


````python
class Camera:

    def take_photo(self):
        print("Photo taken")


class Phone:

    def make_call(self):
        print("Calling...")


class Smartphone(Camera, Phone):
    pass


phone = Smartphone()

phone.take_photo()
phone.make_call()
````

### Important Note About Multiple Inheritance

If multiple parent classes contain methods with the same name, Python must decide which one to use.

Python follows something called the:

> **Method Resolution Order (MRO)**

For now, simply remember:

> Python has an order for searching parent classes.

You can inspect it with:

```python
ClassName.mro()
```


````python
print(Smartphone.mro())
````

# Part 2 — Composition

## 13. Why Do We Need Composition?

Not every relationship makes sense as inheritance.

For example:

```text
Car IS-A Engine
```

That is wrong.

A car is **not** an engine.

Instead:

```text
Car HAS-A Engine
```

The engine is a separate object that the car uses.

That is **composition**.


## 14. What Is Composition?

> **Composition means building a class by using objects of other classes inside it.**

Example:

```text
Car
├── Engine
├── GPS
└── MusicSystem
```

A car **has an** engine.

A car **has a** GPS.

A car **has a** music system.

These parts can each have their own class and behavior.


## 15. How Do We Create Composition?

Create an object of one class inside another class.

Example:

```python
self.engine = Engine()
```

Now the `Car` object contains and uses an `Engine` object.


````python
class Engine:

    def start(self):
        print("Engine started")


class Car:

    def __init__(self):
        # Car HAS-A Engine
        self.engine = Engine()

    def drive(self):
        self.engine.start()
        print("Car is driving")


car = Car()

car.drive()
````

### What happened?

```text
Car Object
    │
    └── engine
          │
          ▼
      Engine Object
          │
          └── start()
```

The `Car` does not inherit from `Engine`.

Instead, the `Car` **contains** an Engine object and uses it.

> **Composition = HAS-A relationship**


## 16. Delegation

In composition, one object often asks another object to do work.

This is called **delegation**.

Example:

```python
self.engine.start()
```

The `Car` says:

> “Engine, you handle starting the engine.”

```text
Car.drive()
    │
    └── delegates engine work
              ↓
        Engine.start()
```


## 17. Why Is Composition Useful?

Composition helps when classes have **different responsibilities**.

Example:

```text
Engine
└── responsible for engine behavior

GPS
└── responsible for navigation

MusicSystem
└── responsible for music

Car
└── combines these objects
```

Instead of making one giant `Car` class do everything, we let smaller classes handle their own jobs.


````python
class Engine:

    def start(self):
        print("Engine started")


class GPS:

    def navigate(self, destination):
        print(f"Navigating to {destination}")


class MusicSystem:

    def play(self):
        print("Playing music")


class Car:

    def __init__(self):
        self.engine = Engine()
        self.gps = GPS()
        self.music = MusicSystem()

    def start_trip(self, destination):
        self.engine.start()
        self.gps.navigate(destination)
        self.music.play()


car = Car()

car.start_trip("San Francisco")
````

## 18. Passing Objects Into Another Class

Composition does not always require creating the object inside the class.

We can also create the object outside and pass it in.

This can make the design more flexible.


````python
class PetrolEngine:

    def start(self):
        print("Petrol engine started")


class ElectricMotor:

    def start(self):
        print("Electric motor started")


class Car:

    def __init__(self, engine):
        self.engine = engine

    def start(self):
        self.engine.start()


petrol_car = Car(PetrolEngine())
electric_car = Car(ElectricMotor())

petrol_car.start()
electric_car.start()
````

### Why is this useful?

`Car` does not care exactly which engine it receives.

It only needs an object that can:

```python
start()
```

So we can replace one component with another more easily.

```text
Car
 │
 ├── PetrolEngine
 │
 or
 └── ElectricMotor
```

This is one reason composition is often very flexible.


# Part 3 — Inheritance and Composition Together

## 19. They Can Be Used in the Same Program

A class can use **inheritance** and **composition** at the same time.

Example:

```text
Smartphone IS-A Device
Smartphone HAS-A Camera
```


````python
class Device:

    def power_on(self):
        print("Device powered on")


class Camera:

    def take_photo(self):
        print("Photo captured")


class Smartphone(Device):

    def __init__(self):
        # Composition
        self.camera = Camera()

    def use_camera(self):
        self.camera.take_photo()


phone = Smartphone()

# Inherited from Device
phone.power_on()

# Uses the composed Camera object
phone.use_camera()
````

### Identify the relationships

```text
Device
   │
   │ IS-A
   ▼
Smartphone
   │
   │ HAS-A
   ▼
 Camera
```

So:

> **Smartphone IS-A Device** → Inheritance

> **Smartphone HAS-A Camera** → Composition


## 20. Inheritance vs Composition

| Inheritance | Composition |
|---|---|
| **IS-A** relationship | **HAS-A** relationship |
| Child gets behavior from parent | Class uses another object |
| Creates a class hierarchy | Combines independent objects |
| Reuse through inheritance | Reuse through contained objects |
| Customize using overriding | Change behavior by replacing components |
| Best for closely related types | Best for combining different responsibilities |

### Examples

```text
Dog IS-A Animal
→ Inheritance

Car HAS-A Engine
→ Composition
```


## 21. How Do I Decide Which One to Use?

Ask one simple question:

### Is the new class truly a type of the other class?

If yes:

```text
Dog IS-A Animal
Manager IS-A Employee
Car IS-A Vehicle
```

Inheritance may make sense.

---

### Does one object simply contain/use another object?

If yes:

```text
Car HAS-A Engine
Computer HAS-A Keyboard
Order HAS-A PaymentMethod
```

Composition may make sense.

### Easy memory

> **IS-A → Inheritance**

> **HAS-A → Composition**

These parts have their own responsibilities and can often be replaced independently.
