# Abstraction in Python

### Simple idea

> **Abstraction = Show what the user needs, hide how it works internally.**

We use abstraction every day.

For example, when you use a payment app, you may simply click:

```text
Pay $100
```

You do not need to know every internal step:

```text
Validate account
→ Contact bank
→ Verify payment
→ Process transaction
→ Update records
→ Return result
```

You only need a simple action:

```python
pay()
```

That is the main idea behind **abstraction**.


## 1. Why Do We Need Abstraction?

Imagine we are building an online shopping app.

Customers can pay using:

- Credit Card
- PayPal
- Bank Transfer

Each payment method works differently internally.

```text
Credit Card
└── validate card → contact bank → process payment

PayPal
└── login → verify account → process payment

Bank Transfer
└── verify bank → transfer money → confirm payment
```

But the shopping app should not need to know all those details.

It should simply be able to say:

```python
payment.pay(100)
```

> The app knows **WHAT to do**: `pay()`  
> The payment class decides **HOW to do it**.


## 2. Without Abstraction

Without a common structure, different payment types may use completely different method names.


````python
class CreditCard:

    def charge_card(self, amount):
        print(f"Charging credit card: ${amount}")


class PayPal:

    def send_paypal_payment(self, amount):
        print(f"Sending PayPal payment: ${amount}")


card = CreditCard()
paypal = PayPal()

card.charge_card(100)
paypal.send_paypal_payment(100)
````

### What is the problem?

Both classes perform the same general job:

> **Make a payment**

But they use different method names:

```text
CreditCard → charge_card()
PayPal     → send_paypal_payment()
```

The rest of our program must remember how every payment type works.

If we later add:

```text
ApplePay
GooglePay
BankTransfer
CryptoPayment
```

the program becomes harder to manage.

Wouldn't it be easier if every payment method followed the same rule?

```text
Every payment method must have:

pay()
```

That is where abstraction helps.


## 3. With Abstraction

We create a common rule:

```text
PaymentMethod
      │
      └── pay()
           │
      every payment type
      must implement it
```

Then:

```text
PaymentMethod
├── CreditCard → pay()
├── PayPal     → pay()
└── BankTransfer → pay()
```

The outside program only needs to know:

```python
payment.pay()
```

It does not need to know the internal payment steps.


## 4. How Do We Achieve Abstraction in Python?

Python provides the `abc` module.

We commonly use:

```python
from abc import ABC, abstractmethod
```

### Two important pieces

- `ABC` → used to create an **Abstract Base Class**
- `@abstractmethod` → marks a method that subclasses **must implement**

### Easy memory

```text
ABC
↓
Creates the common blueprint

@abstractmethod
↓
Creates a rule that child classes must follow
```


## 5. Abstract Class and Abstract Method

An **abstract class** is a class mainly used as a blueprint for other classes.

An **abstract method** says:

> “Every child class must provide its own version of this method.”


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass
````

Here:

```text
PaymentMethod
```

is the **abstract class**.

And:

```text
pay()
```

is the **abstract method**.

We are saying:

> Every payment method must have a `pay()` method.

But we are **not** deciding here how the payment happens.


## 6. Concrete Classes

A normal child class that implements all required abstract methods is called a **concrete class**.

Now each payment type provides its own implementation.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card")


class PayPal(PaymentMethod):

    def pay(self, amount):
        print(f"Paid ${amount} using PayPal")


card = CreditCard()
paypal = PayPal()

card.pay(100)
paypal.pay(200)
````

### What did abstraction give us?

Both payment classes now follow the same interface:

```python
pay()
```

But each class can implement it differently.

```text
CreditCard.pay()
      ↓
Credit card logic

PayPal.pay()
      ↓
PayPal logic
```

The user of the class only needs to remember:

```python
pay()
```

> **Same simple interface, different hidden implementations.**


## 7. Abstract Classes Cannot Be Used Directly

An abstract class is incomplete.

It defines the rules, but some behavior is intentionally left for child classes to implement.

So this is not allowed:

```python
payment = PaymentMethod()
```

Python will raise a `TypeError`.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Uncomment this line to see the error:
# payment = PaymentMethod()

print("We create objects from complete child classes, not from the abstract class.")
````

### Easy way to remember

```text
Abstract Class
      ↓
Blueprint / Rules
      ↓
Usually NOT used to create objects directly

Concrete Child Class
      ↓
Complete implementation
      ↓
Can create objects
```


## 8. What Happens If a Child Class Does NOT Implement the Abstract Method?

If a child class does not implement all required abstract methods, that child class is still incomplete.

Python will not allow us to create an object from it.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class BankTransfer(PaymentMethod):
    pass


# BankTransfer did not implement pay(),
# so this would raise a TypeError:

# bank = BankTransfer()

print("BankTransfer must implement pay() before we can create an object.")
````

## 9. Concrete Methods Inside an Abstract Class

An abstract class can contain **normal methods too**.

These are sometimes called **concrete methods**.

A concrete method already has working code, so child classes can use it directly.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    def payment_message(self):
        print("Thank you for your payment!")


class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Paid ${amount} using Credit Card")


card = CreditCard()

card.pay(100)

# This method came from the abstract parent class
card.payment_message()
````

### Why is this useful?

Some behavior is common to every child class.

Instead of repeating it:

```text
CreditCard → thank you message
PayPal     → thank you message
Bank       → thank you message
```

we can write the common behavior **once** in the abstract parent class.


## 10. Abstract Properties

Abstraction can also require child classes to provide a **property**.

For example, every payment method should provide its own name.

This uses:

```python
@property
@abstractmethod
```

This is useful, but you do not need to memorize it yet.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @property
    @abstractmethod
    def payment_name(self):
        pass

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(PaymentMethod):

    @property
    def payment_name(self):
        return "Credit Card"

    def pay(self, amount):
        print(f"Paid ${amount} using {self.payment_name}")


card = CreditCard()

print(card.payment_name)
card.pay(100)
````

### What is happening?

The abstract class requires every child class to provide:

```text
payment_name
pay()
```

So every payment class follows a consistent structure.

> **Abstract methods require behavior.**  
> **Abstract properties can require data/properties.**


## 11. Real Use Case: One Function Can Work With Different Payment Types

Because every payment class follows the same `pay()` interface, the rest of the program can work with any payment method in the same way.


````python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Credit Card payment: ${amount}")


class PayPal(PaymentMethod):

    def pay(self, amount):
        print(f"PayPal payment: ${amount}")


class BankTransfer(PaymentMethod):

    def pay(self, amount):
        print(f"Bank Transfer payment: ${amount}")


def checkout(payment_method, amount):
    payment_method.pay(amount)


card = CreditCard()
paypal = PayPal()
bank = BankTransfer()

checkout(card, 100)
checkout(paypal, 200)
checkout(bank, 300)
````

### This is the power of abstraction

The `checkout()` function does **not care** whether the payment is:

- Credit Card
- PayPal
- Bank Transfer

It only knows:

```python
payment_method.pay(amount)
```

The complicated details are hidden inside each payment class.

```text
checkout()
    │
    └── pay()
         │
         ├── Credit Card logic
         ├── PayPal logic
         └── Bank Transfer logic
```

> The outside code works with a **simple common interface**.


## 12. Another Simple Example: Shapes

Every shape should be able to calculate its area.

But the calculation is different:

```text
Circle    → π × radius²
Rectangle → length × width
Triangle  → ½ × base × height
```

We can define one rule:

```text
Every Shape must have area()
```


````python
from abc import ABC, abstractmethod
import math


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle area:", circle.area())
print("Rectangle area:", rectangle.area())
````

## 14. Abstraction vs Encapsulation

These two concepts are related, but they are not the same.

| Encapsulation | Abstraction |
|---|---|
| Controls access to data | Hides unnecessary implementation details |
| Protects internal state | Shows a simple interface |
| Example: `__balance` | Example: `pay()` |
| Focus: **protect/control data** | Focus: **what to use, not how it works** |

### Easy memory

> **Encapsulation = Protect it**

> **Abstraction = Simplify it**
